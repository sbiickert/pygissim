```python
import ipywidgets as widgets
from IPython.display import display
```

```python
def createDialog():
    text = widgets.Text(placeholder="Enter Text", description="Text:")
    dropdown = widgets.Dropdown(options=['apple','banana','cherry'], value='banana', description="Fruit:")
    slider = widgets.IntSlider(min=1, max=10, value=7, step=1, description="Number:")
    okButton = widgets.Button(description='OK', button_style='success')
    cancelButton = widgets.Button(description='Cancel', button_style='')
    hbox = widgets.HBox([okButton, cancelButton])
    vbox = widgets.VBox([text, dropdown, slider, hbox])
    return vbox
```

```python
w = createDialog()
display(w)
```

```python
w.close()
```
